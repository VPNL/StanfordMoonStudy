## Section 3: Visual Angle Helper Function

```r
observer_height = 102.5

# Calculate visual angle helper function
calc_va_df <- function(df, size_col, distance_col) {
  if (is.character(size_col)) {
    size <- df[[size_col]]
  } else {
    size <- size_col
  }
  dist <- df[[distance_col]]
  (2 * atan(size / (2 * dist))) * (180 / pi)
}

calc_observer_distance_df <- function(data, distance_col, height_col) {

  distance <- data[[distance_col]]
  if (is.character(height_col)) {
    object_height <- data[[height_col]]
  } else {
    object_height <- height_col
  }

  height_diff <- object_height - observer_height

  sqrt(distance^2 + height_diff^2)
}

calc_observer_elevation_df <- function(data, distance_col, height_col) {

  distance <- data[[distance_col]]
  if (is.character(height_col)) {
    object_height <- data[[height_col]]
  } else {
    object_height <- height_col
  }

  height_diff <- object_height - observer_height

  atan(height_diff / distance) * 180 / pi
}
```
