## Section 2: Load Data


```r
# Upload Processed Data Sheet
data_long <- read.csv("Data.csv")
```

```r
# Upload Stereotest Data Here
data <- read.csv("StereoData.csv", skip=2)
print(data)
```

```r
data_long <- data_long %>%
  left_join(
    data %>%
      select(
        ID,
        `Stereoscore Circles` = Stereoscore.Circles,
        `Stereoscore Animals` = Stereoscore.Animals,
        `Stereoscore Randot` = Stereoscore.Randot,
        `NormedScore` = Normed.Score,
        `Stereovision_arcsec` = Stereovision_arcseconds,
        `Clinical Notes` = Clinical.Notes,
      ),
    by = "ID"
  )
```

```r
write.csv(data_long, file = "QuadProcessedParallaxandStereoData.csv")
```
